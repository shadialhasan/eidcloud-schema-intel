# 🗄️ eidcloud-schema-intel Quickstart

Welcome to **eidcloud-schema-intel**! This notebook demonstrates how to inspect an unindexed SQLite database, automatically infer hidden/implicit foreign keys, detect missing indexes, and render a Mermaid ER diagram—all powered by pure PHP 8.2+ without any external dependencies.

```python
# Setup: Ensure PHP is installed and clone eidcloud-schema-intel
!apt-get update -qq && apt-get install -y -qq php-cli php-sqlite3 > /dev/null 2>&1
!git clone https://github.com/mhds-alhasan/eidcloud-schema-intel.git /content/eidcloud-schema-intel 2>/dev/null || true
%cd /content/eidcloud-schema-intel
!php -v
```

```python
# Create sample SQLite database with missing indexes and implicit relations
import sqlite3

conn = sqlite3.connect('demo.sqlite')
cursor = conn.cursor()
cursor.executescript('''
CREATE TABLE users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    email TEXT NOT NULL,
    name TEXT NOT NULL,
    status TEXT NOT NULL
);

CREATE TABLE posts (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER NOT NULL,
    title TEXT NOT NULL,
    content TEXT,
    created_by INTEGER
);

CREATE TABLE comments (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    post_id INTEGER NOT NULL,
    author_id INTEGER NOT NULL,
    body TEXT NOT NULL
);
''')
conn.commit()
conn.close()
print("Sample database 'demo.sqlite' created successfully.")
```

```python
# Inspect schema, discover implicit relations, detect missing indexes, and render Mermaid ER
print("--- 1. Missing Indexes Advisory ---")
!php bin/eidcloud-schema missing-indexes --sqlite=demo.sqlite

print("\n--- 2. Inferred Mermaid ER Diagram ---")
!php bin/eidcloud-schema mermaid --sqlite=demo.sqlite
```
